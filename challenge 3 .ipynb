```python

# ============================================================
# CHALLENGE 3: CAN YOU CHANGE THE CORRELATION?
# Quantum Computing Assignment using Qiskit
# ============================================================

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# Initialize quantum simulator
simulator = AerSimulator()

# ------------------------------------------------------------
# 1. ORIGINAL BELL-STATE CIRCUIT
# ------------------------------------------------------------

original = QuantumCircuit(2, 2)

# Create superposition on qubit 0
original.h(0)

# Entangle qubit 0 and qubit 1
original.cx(0, 1)

# Measure both qubits
original.measure([0, 1], [0, 1])

print("ORIGINAL BELL-STATE CIRCUIT")
print(original.draw())

# ------------------------------------------------------------
# 2. MODIFIED CIRCUIT
# Add exactly ONE extra quantum gate: X on qubit 0
# ------------------------------------------------------------

modified = QuantumCircuit(2, 2)

# Original Bell-state operations
modified.h(0)
modified.cx(0, 1)

# Modify the correlation using one X gate
modified.x(0)

# Measure both qubits
modified.measure([0, 1], [0, 1])

print("\nMODIFIED QUANTUM CIRCUIT")
print(modified.draw())

# ------------------------------------------------------------
# 3. PREDICTION BEFORE EXECUTION
# ------------------------------------------------------------

print("\nPREDICTION")
print("Original circuit: Outcomes 00 and 11 are expected.")
print("Modified circuit: Outcomes 01 and 10 are expected.")
print("Each expected outcome has approximately 50% probability.")

# ------------------------------------------------------------
# 4. EXECUTE BOTH CIRCUITS FOR 100 SHOTS
# ------------------------------------------------------------

original_result = simulator.run(
    original, shots=100
).result()

modified_result = simulator.run(
    modified, shots=100
).result()

original_counts = original_result.get_counts()
modified_counts = modified_result.get_counts()

# ------------------------------------------------------------
# 5. DISPLAY MEASUREMENT RESULTS
# ------------------------------------------------------------

print("\nORIGINAL CIRCUIT RESULTS")
print("Measurement counts:", original_counts)
print("Total shots:", sum(original_counts.values()))

print("\nMODIFIED CIRCUIT RESULTS")
print("Measurement counts:", modified_counts)
print("Total shots:", sum(modified_counts.values()))

# ------------------------------------------------------------
# 6. DISPLAY HISTOGRAMS
# ------------------------------------------------------------

plot_histogram(
    original_counts,
    title="Original Bell State - 100 Shots",
    color=["royalblue"]
)
plt.show()

plot_histogram(
    modified_counts,
    title="Modified Bell State - 100 Shots",
    color=["darkorange"]
)
plt.show()

# ------------------------------------------------------------
# 7. COMPARE BOTH CIRCUITS
# ------------------------------------------------------------

print("\nCOMPARISON OF MEASUREMENT OUTCOMES")

all_outcomes = ["00", "01", "10", "11"]

print(f"{'Outcome':<12}{'Original':<15}{'Modified':<15}")

for outcome in all_outcomes:
    original_count = original_counts.get(outcome, 0)
    modified_count = modified_counts.get(outcome, 0)

    print(
        f"{outcome:<12}"
        f"{original_count:<15}"
        f"{modified_count:<15}"
    )

# ------------------------------------------------------------
# 8. FINAL EXPLANATION
# ------------------------------------------------------------

print("\nFINAL EXPLANATION")
print(
    "The original Bell-state circuit uses an H gate and a CX gate "
    "to create an entangled state. Its measurement outcomes are "
    "00 and 11, each with approximately 50% probability."
)

print(
    "Adding one X gate to qubit 0 flips that qubit. The modified "
    "circuit produces outcomes 01 and 10, each with approximately "
    "50% probability. The outcomes change, but the qubits remain "
    "perfectly correlated or anticorrelated in the computational "
    "basis. Small count differences can occur due to sampling."
)
```
